```python
def detect_outliers_iqr(

    data,

    column
):

    Q1 = data[column].quantile(

        0.25
    )


    Q3 = data[column].quantile(

        0.75
    )


    IQR = Q3 - Q1


    lower_bound = (

        Q1

        -

        1.5 * IQR
    )


    upper_bound = (

        Q3

        +

        1.5 * IQR
    )


    outliers = (

        (data[column] < lower_bound)

        |

        (data[column] > upper_bound)
    )


    return (

        outliers,

        lower_bound,

        upper_bound
    )
```

```python
for column in sensor_columns:

    outliers, lower, upper = (

        detect_outliers_iqr(

            df,

            column
        )
    )


    print(
        "\n" + "=" * 50
    )


    print(
        "Sensor:",
        column
    )


    print(
        "Lower Bound:",
        lower
    )


    print(
        "Upper Bound:",
        upper
    )


    print(
        "Possible Outliers:",
        outliers.sum()
    )
```

```python
clean_sensor_columns = []
for column in sensor_columns:

    clean_column = (

        column

        + "_Clean"
    )


    clean_sensor_columns.append(

        clean_column
    )


    df[clean_column] = (

        df[column]

        .copy()
    )
```

```python
for column in sensor_columns:

    clean_column = (

        column

        + "_Clean"
    )


    outliers, _, _ = (

        detect_outliers_iqr(

            df,

            column
        )
    )


    df.loc[

        outliers,

        clean_column

    ] = np.nan
```

```python
for column in clean_sensor_columns:

    df[column] = (

        df[column]

        .interpolate(

            method="linear"
        )

        .ffill()

        .bfill()
    )
```

```python
SMOOTHING_WINDOW = 5
```

```python
df["Temperature_Smoothed"] = (

    df["Temperature_Clean"]

    .rolling(

        window=SMOOTHING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Humidity_Smoothed"] = (

    df["Humidity_Clean"]

    .rolling(

        window=SMOOTHING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Gas_Smoothed"] = (

    df["Gas_Clean"]

    .rolling(

        window=SMOOTHING_WINDOW,

        min_periods=1
    )

    .mean()
)
```
