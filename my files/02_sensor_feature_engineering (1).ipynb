```python
display(
    df.head()
)
```

```python
print("AVAILABLE COLUMNS:\n")

for column in df.columns:

    print("-", column)
```

```python
df["Timestamp"] = pd.to_datetime(

    df["Timestamp"],

    errors="coerce"
)
```

```python
df = df.sort_values(

    by="Timestamp"
)
df = df.reset_index(

    drop=True
)
```

```python
TEMPERATURE_COLUMN = None
HUMIDITY_COLUMN = None
GAS_COLUMN = None
```

```python
if "Temperature_Smoothed" in df.columns:

    TEMPERATURE_COLUMN = "Temperature_Smoothed"

else:

    TEMPERATURE_COLUMN = "Temperature"
```

```python
if "Humidity_Smoothed" in df.columns:

    HUMIDITY_COLUMN = "Humidity_Smoothed"

else:

    HUMIDITY_COLUMN = "Humidity"
```

```python
if "Gas_Smoothed" in df.columns:

    GAS_COLUMN = "Gas_Smoothed"

else:

    GAS_COLUMN = "Gas"
```

```python
print("Selected Sensor Columns:\n")

print(
    "Temperature:",
    TEMPERATURE_COLUMN
)

print(
    "Humidity:",
    HUMIDITY_COLUMN
)

print(
    "Gas:",
    GAS_COLUMN
)
```

```python
df["Temperature_Feature"] = df[
    TEMPERATURE_COLUMN
]

df["Humidity_Feature"] = df[
    HUMIDITY_COLUMN
]

df["Gas_Feature"] = df[
    GAS_COLUMN
]
```

```python
#time features
```

```python
df["Year"] = (

    df["Timestamp"]

    .dt.year
)
df["Month"] = (

    df["Timestamp"]

    .dt.month
)
df["Day_Of_Month"] = (

    df["Timestamp"]

    .dt.day
)
df["Hour"] = (

    df["Timestamp"]

    .dt.hour
)
df["Minute"] = (

    df["Timestamp"]

    .dt.minute
)
```

```python
df["Day_Of_Week"] = (

    df["Timestamp"]

    .dt.dayofweek
)
```

```python
start_time = df[
    "Timestamp"
].min()
df["Elapsed_Time_Hours"] = (

    df["Timestamp"]

    -

    start_time
)
df["Elapsed_Time_Hours"] = (

    df["Elapsed_Time_Hours"]

    .dt.total_seconds()

    /

    3600
)
```

```python
df["Elapsed_Time_Days"] = (

    df["Elapsed_Time_Hours"]

    /

    24
)
```

```python
#SENSOR CHANGE FEATURES
```

```python
df["Temperature_Change"] = (

    df["Temperature_Feature"]

    .diff()
)
```

```python
df["Humidity_Change"] = (

    df["Humidity_Feature"]

    .diff()
)
```

```python
df["Gas_Change"] = (

    df["Gas_Feature"]

    .diff()
)
```

```python
change_columns = [

    "Temperature_Change",

    "Humidity_Change",

    "Gas_Change"
]
df[change_columns] = (

    df[change_columns]

    .fillna(
        0
    )
)
```

```python
#RATE OF CHANGE FEATURES
```

```python
df["Time_Difference_Hours"] = (

    df["Timestamp"]

    .diff()

    .dt.total_seconds()

    /

    3600
)
```

```python
df["Time_Difference_Hours"] = (

    df["Time_Difference_Hours"]

    .replace(
        0,
        np.nan
    )
)
```

```python
df["Temperature_Rate"] = (

    df["Temperature_Change"]

    /

    df["Time_Difference_Hours"]
)
```

```python
df["Humidity_Rate"] = (

    df["Humidity_Change"]

    /

    df["Time_Difference_Hours"]
)
```

```python
df["Gas_Rate"] = (

    df["Gas_Change"]

    /

    df["Time_Difference_Hours"]
)
```

```python
rate_columns = [

    "Temperature_Rate",

    "Humidity_Rate",

    "Gas_Rate"
]
df[rate_columns] = (

    df[rate_columns]

    .replace(

        [

            np.inf,

            -np.inf

        ],

        np.nan
    )

    .fillna(
        0
    )
)
```

```python
#ROLLING FEATURES
```

```python
ROLLING_WINDOW = 5
```

```python
df["Temperature_Rolling_Mean"] = (

    df["Temperature_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Humidity_Rolling_Mean"] = (

    df["Humidity_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Gas_Rolling_Mean"] = (

    df["Gas_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Temperature_Rolling_STD"] = (

    df["Temperature_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .std()

    .fillna(
        0
    )
)
```

```python
df["Humidity_Rolling_Mean"] = (

    df["Humidity_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Gas_Rolling_Mean"] = (

    df["Gas_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .mean()
)
```

```python
df["Temperature_Rolling_STD"] = (

    df["Temperature_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .std()

    .fillna(
        0
    )
)
```

```python
df["Humidity_Rolling_STD"] = (

    df["Humidity_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .std()

    .fillna(
        0
    )
)
```

```python
df["Gas_Rolling_STD"] = (

    df["Gas_Feature"]

    .rolling(

        window=ROLLING_WINDOW,

        min_periods=1
    )

    .std()

    .fillna(
        0
    )
)
```

```python
#CELL 35 — Temperature × Humidity
df["Temperature_Humidity_Interaction"] = (

    df["Temperature_Feature"]

    *

    df["Humidity_Feature"]
)
```

```python
df["Temperature_Gas_Interaction"] = (

    df["Temperature_Feature"]

    *

    df["Gas_Feature"]
)
```

```python
df["Humidity_Gas_Interaction"] = (

    df["Humidity_Feature"]

    *

    df["Gas_Feature"]
)
```

```python
#CUMULATIVE EXPOSURE FEATURES
df["Cumulative_Temperature"] = (

    df["Temperature_Feature"]

    .cumsum()
)
```

```python
df["Cumulative_Humidity"] = (

    df["Humidity_Feature"]

    .cumsum()
)
```

```python
df["Cumulative_Gas"] = (

    df["Gas_Feature"]

    .cumsum()
)
```

```python
#NORMALIZED CUMULATIVE EXPOSURE
initial_temperature = df[
    "Temperature_Feature"
].iloc[0]
df["Temperature_From_Initial"] = (

    df["Temperature_Feature"]

    -

    initial_temperature
)
```

```python
initial_humidity = df[
    "Humidity_Feature"
].iloc[0]
df["Humidity_From_Initial"] = (

    df["Humidity_Feature"]

    -

    initial_humidity
)
```

```python
initial_gas = df[
    "Gas_Feature"
].iloc[0]
df["Gas_From_Initial"] = (

    df["Gas_Feature"]

    -

    initial_gas
)
```

```python
#CELL 44 — Gas Increase from Previous Reading
df["Gas_Increasing"] = (

    df["Gas_Change"]

    >

    0
).astype(
    int
)
```

```python
df["Temperature_Increasing"] = (

    df["Temperature_Change"]

    >

    0
).astype(
    int
)
```

```python
df["Humidity_Increasing"] = (

    df["Humidity_Change"]

    >

    0
).astype(
    int
)
```

```python
#SENSOR TREND FEATURES
TREND_WINDOW = 5
```

```python
df["Temperature_Trend"] = (

    df["Temperature_Feature"]

    -

    df["Temperature_Rolling_Mean"]
)
```

```python
df["Humidity_Trend"] = (

    df["Humidity_Feature"]

    -

    df["Humidity_Rolling_Mean"]
)
```

```python
df["Gas_Trend"] = (

    df["Gas_Feature"]

    -

    df["Gas_Rolling_Mean"]
)
```

```python
#CATEGORICAL ENVIRONMENT FEATURES
df["Temperature_Level"] = pd.cut(

    df["Temperature_Feature"],

    bins=[
        -np.inf,
        10,
        20,
        30,
        np.inf
    ],

    labels=[
        "Cold",
        "Cool",
        "Moderate",
        "Hot"
    ]
)
```

```python
df["Humidity_Level"] = pd.cut(

    df["Humidity_Feature"],

    bins=[
        -np.inf,
        30,
        60,
        80,
        np.inf
    ],

    labels=[
        "Low",
        "Moderate",
        "High",
        "Very_High"
    ]
)
```

```python
gas_q1 = df[
    "Gas_Feature"
].quantile(
    0.25
)

gas_q2 = df[
    "Gas_Feature"
].quantile(
    0.50
)

gas_q3 = df[
    "Gas_Feature"
].quantile(
    0.75
)
```

```python
# ============================================================
# GAS LEVEL FEATURE
# ============================================================

# Check how many unique gas values exist
unique_gas_values = df["Gas_Feature"].nunique()

print("Unique Gas Values:", unique_gas_values)


if unique_gas_values >= 4:

    # Rank the values first.
    # This prevents duplicate quantile edges from causing errors.
    gas_rank = df["Gas_Feature"].rank(
        method="first"
    )

    df["Gas_Level"] = pd.qcut(
        gas_rank,
        q=4,
        labels=[
            "Low",
            "Medium",
            "High",
            "Very_High"
        ]
    )


elif unique_gas_values == 3:

    df["Gas_Level"] = pd.cut(
        df["Gas_Feature"],
        bins=3,
        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


elif unique_gas_values == 2:

    df["Gas_Level"] = pd.cut(
        df["Gas_Feature"],
        bins=2,
        labels=[
            "Low",
            "High"
        ]
    )


else:

    # All gas values are identical
    df["Gas_Level"] = "Constant"


print("\nGas Level created successfully.")

print(
    df["Gas_Level"]
    .value_counts(
        dropna=False
    )
)
```

```python
#FINAL DATA CLEANING
```

```python
df = df.replace(

    [

        np.inf,

        -np.inf

    ],

    np.nan
)
```

```python
missing_values = (

    df

    .isnull()

    .sum()
)
print(
    "MISSING VALUES:\n"
)

print(
    missing_values[
        missing_values > 0
    ]
)
```

```python
numeric_columns = df.select_dtypes(

    include=[
        np.number
    ]

).columns
for column in numeric_columns:

    df[column] = (

        df[column]

        .ffill()

        .bfill()
    )
```
