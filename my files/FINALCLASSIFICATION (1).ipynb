```python
from pathlib import Path
from PIL import Image

DATASET_ROOT = Path(
    "/content/drive/MyDrive/CAPSTONE DATASET/"
    "external_dataset/banana_classification_model2/"
    "banana_classification"
)

CLASSES = ["unripe", "ripe", "overripe", "rotten"]
SPLITS = ["train", "valid", "test"]

print("Dataset verification\n")

for split in SPLITS:
    print(f"{split.upper()}")

    for cls in CLASSES:
        folder = DATASET_ROOT / split / cls
        images = list(folder.glob("*.jpg")) + list(folder.glob("*.jpeg")) + list(folder.glob("*.png"))

        print(f"  {cls:10s}: {len(images)} images")

        # Check one image can be opened
        if images:
            try:
                with Image.open(images[0]) as img:
                    print(f"             sample size: {img.size}, mode: {img.mode}")
            except Exception as e:
                print(f"             ERROR reading image: {e}")

    print()
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

TRAIN_DIR = str(DATASET_ROOT / "train")
VAL_DIR   = str(DATASET_ROOT / "valid")
TEST_DIR  = str(DATASET_ROOT / "test")

CLASS_NAMES = ["unripe", "ripe", "overripe", "rotten"]

train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    labels="inferred",
    label_mode="int",
    class_names=CLASS_NAMES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    labels="inferred",
    label_mode="int",
    class_names=CLASS_NAMES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    labels="inferred",
    label_mode="int",
    class_names=CLASS_NAMES,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("\nClass names:", CLASS_NAMES)
print("Number of classes:", len(CLASS_NAMES))

print("\nBatches:")
print("Train:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation:", tf.data.experimental.cardinality(val_ds).numpy())
print("Test:", tf.data.experimental.cardinality(test_ds).numpy())
```

```python
# Training-only data augmentation

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.08),
    tf.keras.layers.RandomTranslation(0.08, 0.08),
    tf.keras.layers.RandomZoom(
        height_factor=(-0.10, 0.10),
        width_factor=(-0.10, 0.10)
    ),
    tf.keras.layers.RandomContrast(0.10),
    tf.keras.layers.RandomBrightness(0.10),
], name="training_augmentation")

print(data_augmentation.summary())
```

```python
# Model 2 — MobileNetV2 baseline

NUM_CLASSES = 4

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

inputs = tf.keras.Input(shape=(224, 224, 3), name="image")

x = data_augmentation(inputs)
x = tf.keras.applications.mobilenet_v2.preprocess_input(x)

x = base_model(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)

x = tf.keras.layers.Dropout(0.30)(x)
x = tf.keras.layers.Dense(128, activation="relu")(x)
x = tf.keras.layers.Dropout(0.30)(x)

outputs = tf.keras.layers.Dense(
    NUM_CLASSES,
    activation="softmax",
    name="ripeness_output"
)(x)

model2 = tf.keras.Model(inputs, outputs, name="banana_ripeness_model2")

model2.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model2.summary()
```

```python
# ============================================================
# MODEL 2 — CONTROLLED FINE-TUNING EXPERIMENT
# Baseline remains untouched
# ============================================================

import tensorflow as tf
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.metrics import classification_report, confusion_matrix

print("=" * 60)
print("MODEL 2 — CONTROLLED FINE-TUNING")
print("=" * 60)

# ------------------------------------------------------------
# 1. Load the BEST BASELINE model
# ------------------------------------------------------------
BASELINE_PATH = Path(
    "/content/drive/MyDrive/CAPSTONE DATASET/mmodels/"
    "banana_ripeness_model2_best.keras"
)

finetune_model = tf.keras.models.load_model(BASELINE_PATH)

print("\nBaseline model loaded:")
print(BASELINE_PATH)

# ------------------------------------------------------------
# 2. Find MobileNetV2 backbone
# ------------------------------------------------------------
backbone = None

for layer in finetune_model.layers:
    if isinstance(layer, tf.keras.Model) and "mobilenetv2" in layer.name.lower():
        backbone = layer
        break

if backbone is None:
    raise ValueError("MobileNetV2 backbone could not be found.")

print("Backbone:", backbone.name)
print("Total backbone layers:", len(backbone.layers))

# ------------------------------------------------------------
# 3. Freeze everything first
# ------------------------------------------------------------
backbone.trainable = True

# Fine-tune only the upper part of MobileNetV2.
# Keep the lower feature-extraction layers frozen.
FINE_TUNE_FROM = len(backbone.layers) - 30

for layer in backbone.layers[:FINE_TUNE_FROM]:
    layer.trainable = False

# Keep Batch Normalization layers frozen.
for layer in backbone.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

print("Fine-tuning from backbone layer:", FINE_TUNE_FROM)
print("Trainable backbone layers:",
      sum(layer.trainable for layer in backbone.layers))

# ------------------------------------------------------------
# 4. Recompile with VERY small learning rate
# ------------------------------------------------------------
finetune_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nFine-tuning configuration:")
print("Learning rate: 1e-5")
print("Upper MobileNetV2 layers fine-tuned: 30")
print("Batch Normalization layers: frozen")

# ------------------------------------------------------------
# 5. Save fine-tuned model separately
# ------------------------------------------------------------
MODEL2_RESULTS_DIR = Path(
    "/content/drive/MyDrive/CAPSTONE DATASET/mmodels/model2_results"
)
MODEL2_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

FINETUNE_BEST_PATH = (
    MODEL2_RESULTS_DIR /
    "banana_ripeness_model2_finetuned_best.keras"
)

callbacks_finetune = [
    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(FINETUNE_BEST_PATH),
        monitor="val_accuracy",
        mode="max",
        save_best_only=True,
        verbose=1
    ),

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        mode="min",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        mode="min",
        factor=0.3,
        patience=1,
        min_lr=1e-7,
        verbose=1
    )
]

# ------------------------------------------------------------
# 6. Fine-tune
# ------------------------------------------------------------
print("\nStarting fine-tuning...\n")

history_finetune = finetune_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks_finetune,
    verbose=1
)

# ------------------------------------------------------------
# 7. Evaluate fine-tuned model on untouched TEST set
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("FINE-TUNED MODEL — TEST EVALUATION")
print("=" * 60)

test_loss_ft, test_accuracy_ft = finetune_model.evaluate(
    test_ds,
    verbose=1
)

# ------------------------------------------------------------
# 8. Generate detailed predictions
# ------------------------------------------------------------
y_true_ft = []
y_pred_ft = []

for images, labels in test_ds:
    predictions = finetune_model.predict(
        images,
        verbose=0
    )

    y_true_ft.extend(labels.numpy())
    y_pred_ft.extend(np.argmax(predictions, axis=1))

y_true_ft = np.array(y_true_ft)
y_pred_ft = np.array(y_pred_ft)

# ------------------------------------------------------------
# 9. Classification report
# ------------------------------------------------------------
report_ft = classification_report(
    y_true_ft,
    y_pred_ft,
    target_names=CLASS_NAMES,
    digits=4,
    output_dict=True
)

report_df_ft = pd.DataFrame(report_ft).transpose()

print("\nClassification Report")
print("=====================")

print(
    classification_report(
        y_true_ft,
        y_pred_ft,
        target_names=CLASS_NAMES,
        digits=4
    )
)

# ------------------------------------------------------------
# 10. Confusion matrix
# ------------------------------------------------------------
cm_ft = confusion_matrix(
    y_true_ft,
    y_pred_ft
)

print("\nConfusion Matrix")
print("================")
print("Rows = True class")
print("Columns = Predicted class")
print(CLASS_NAMES)
print(cm_ft)

# ------------------------------------------------------------
# 11. Compare baseline vs fine-tuned model
# ------------------------------------------------------------
baseline_accuracy = test_accuracy
baseline_macro_f1 = report_dict["macro avg"]["f1-score"]

finetune_macro_f1 = report_ft["macro avg"]["f1-score"]

accuracy_change = (
    test_accuracy_ft - baseline_accuracy
) * 100

f1_change = (
    finetune_macro_f1 - baseline_macro_f1
) * 100

print("\n" + "=" * 60)
print("BASELINE vs FINE-TUNED")
print("=" * 60)

print(f"Baseline Test Accuracy:     {baseline_accuracy:.4%}")
print(f"Fine-tuned Test Accuracy:   {test_accuracy_ft:.4%}")
print(f"Accuracy Change:             {accuracy_change:+.2f} percentage points")

print(f"\nBaseline Macro F1:           {baseline_macro_f1:.4f}")
print(f"Fine-tuned Macro F1:         {finetune_macro_f1:.4f}")
print(f"Macro F1 Change:             {f1_change:+.2f} percentage points")

print(f"\nBaseline Test Loss:           {test_loss:.4f}")
print(f"Fine-tuned Test Loss:         {test_loss_ft:.4f}")

# ------------------------------------------------------------
# 12. Save fine-tuned evaluation files
# ------------------------------------------------------------
ft_report_path = (
    MODEL2_RESULTS_DIR /
    "model2_finetuned_classification_report.csv"
)

ft_cm_path = (
    MODEL2_RESULTS_DIR /
    "model2_finetuned_confusion_matrix.csv"
)

ft_summary_path = (
    MODEL2_RESULTS_DIR /
    "model2_finetuned_test_summary.csv"
)

report_df_ft.to_csv(ft_report_path)

cm_df_ft = pd.DataFrame(
    cm_ft,
    index=[f"True_{c}" for c in CLASS_NAMES],
    columns=[f"Pred_{c}" for c in CLASS_NAMES]
)

cm_df_ft.to_csv(ft_cm_path)

summary_ft = pd.DataFrame([{
    "baseline_test_accuracy": baseline_accuracy,
    "finetuned_test_accuracy": test_accuracy_ft,
    "accuracy_change_percentage_points": accuracy_change,

    "baseline_macro_f1": baseline_macro_f1,
    "finetuned_macro_f1": finetune_macro_f1,
    "macro_f1_change": f1_change,

    "baseline_test_loss": test_loss,
    "finetuned_test_loss": test_loss_ft,

    "test_images": len(y_true_ft),
    "misclassified": int(
        np.sum(y_true_ft != y_pred_ft)
    )
}])

summary_ft.to_csv(
    ft_summary_path,
    index=False
)

# ------------------------------------------------------------
# 13. Final decision
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("FINAL MODEL DECISION")
print("=" * 60)

if test_accuracy_ft > baseline_accuracy:
    print("Fine-tuning IMPROVED test accuracy.")
    print("Fine-tuned model should be considered for Model 2.")
elif test_accuracy_ft < baseline_accuracy:
    print("Fine-tuning REDUCED test accuracy.")
    print("Keep the original baseline Model 2.")
else:
    print("Fine-tuning produced the SAME test accuracy.")
    print("Keep the simpler baseline unless Macro F1 improved.")

print("\nSaved fine-tuned model:")
print(FINETUNE_BEST_PATH)

print("\nSaved evaluation files:")
print(ft_report_path)
print(ft_cm_path)
print(ft_summary_path)

print("\n" + "=" * 60)
print("FINE-TUNING EXPERIMENT COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# BANANA CLASSIFICATION — UPLOAD + LIVE WEBCAM INTERFACE
# Model 1 + Model 2
# ============================================================

!pip -q install gradio

import tensorflow as tf
import numpy as np
import gradio as gr
from pathlib import Path

# ------------------------------------------------------------
# MODEL PATHS
# ------------------------------------------------------------

MODEL1_PATH = (
    "/content/drive/MyDrive/CAPSTONE DATASET/mmodels/"
    "banana_ripeness_4class.keras"
)

MODEL2_PATH = (
    "/content/drive/MyDrive/CAPSTONE DATASET/mmodels/"
    "model2_results/banana_ripeness_model2_finetuned_best.keras"
)

# ------------------------------------------------------------
# CLASS DEFINITIONS
# ------------------------------------------------------------

MODEL1_CLASSES = [
    "Green",
    "Overripe",
    "Ripe",
    "Semi-ripe"
]

MODEL2_CLASSES = [
    "Unripe",
    "Ripe",
    "Overripe",
    "Rotten"
]

# ------------------------------------------------------------
# LOAD MODELS
# ------------------------------------------------------------

print("Loading Model 1...")
model1 = tf.keras.models.load_model(MODEL1_PATH)

print("Loading Model 2...")
model2 = tf.keras.models.load_model(MODEL2_PATH)

print("Both classification models loaded successfully.")

# ------------------------------------------------------------
# PREDICTION FUNCTION
# ------------------------------------------------------------

def predict_banana(image, selected_model):

    if image is None:
        return "Please upload an image or capture one using the webcam.", None

    # Select model and class names
    if selected_model == "Model 1 — BananaImageBD":
        model = model1
        class_names = MODEL1_CLASSES

    else:
        model = model2
        class_names = MODEL2_CLASSES

    # Convert image to TensorFlow format
    image = np.array(image)

    # Ensure RGB
    if image.ndim == 2:
        image = np.stack([image] * 3, axis=-1)

    if image.shape[-1] == 4:
        image = image[:, :, :3]

    # Resize to model input
    image = tf.image.resize(
        image,
        (224, 224)
    )

    # Convert to float
    image = tf.cast(
        image,
        tf.float32
    )

    # Add batch dimension
    image = tf.expand_dims(
        image,
        axis=0
    )

    # Model prediction
    probabilities = model.predict(
        image,
        verbose=0
    )[0]

    predicted_index = int(
        np.argmax(probabilities)
    )

    predicted_class = class_names[
        predicted_index
    ]

    confidence = float(
        probabilities[predicted_index]
    )

    # --------------------------------------------------------
    # Confidence table
    # --------------------------------------------------------

    confidence_text = "\n".join(
        [
            f"{class_names[i]:12s}: "
            f"{probabilities[i] * 100:.2f}%"
            for i in range(len(class_names))
        ]
    )

    result_text = (
        f"Prediction: {predicted_class}\n"
        f"Confidence: {confidence * 100:.2f}%\n\n"
        f"Class probabilities:\n"
        f"{confidence_text}"
    )

    # Data for bar chart
    probability_dict = {
        class_names[i]: float(probabilities[i])
        for i in range(len(class_names))
    }

    return result_text, probability_dict


# ------------------------------------------------------------
# GRADIO INTERFACE
# ------------------------------------------------------------

with gr.Blocks(
    title="Banana Classification Interface"
) as demo:

    gr.Markdown(
        """
        # Banana Ripeness / Condition Classifier

        Upload a banana image or use the webcam for live input.

        **Model 1:** Green / Overripe / Ripe / Semi-ripe

        **Model 2:** Unripe / Ripe / Overripe / Rotten
        """
    )

    with gr.Row():

        with gr.Column():

            model_selector = gr.Radio(
                choices=[
                    "Model 1 — BananaImageBD",
                    "Model 2 — External Dataset"
                ],
                value="Model 2 — External Dataset",
                label="Select Classification Model"
            )

            image_input = gr.Image(
                type="numpy",
                sources=["upload", "webcam"],
                label="Upload Test Image or Use Webcam"
            )

            predict_button = gr.Button(
                "Classify Banana",
                variant="primary"
            )

            clear_button = gr.ClearButton(
                components=[image_input]
            )

        with gr.Column():

            result_output = gr.Textbox(
                label="Prediction Result",
                lines=10
            )

            probability_output = gr.Label(
                label="Class Probabilities",
                num_top_classes=4
            )

    predict_button.click(
        fn=predict_banana,
        inputs=[
            image_input,
            model_selector
        ],
        outputs=[
            result_output,
            probability_output
        ]
    )

print("\nStarting interface...")
print("Upload an image or select Webcam in the image box.")

demo.launch(
    share=True,
    debug=True
)
```
